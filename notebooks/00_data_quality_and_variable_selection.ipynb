# WVS Wave 7 – Variabelurval och datakvalitet

**Syfte:** avgränsa ett preliminärt europeiskt urval, välja projektvariabler och undersöka datakvalitet. Detta är en första kontroll; landurval och specialkoder ska bekräftas mot kodboken och diskuteras med läraren innan slutlig analys.

**Forskningsfråga:** I vilken utsträckning kan ekonomiska, sociala och personliga faktorer bidra till att förklara skillnader i människors livstillfredsställelse?

**Målvariabel:** Q49 (Life satisfaction). Q288 (Household income) används som prediktor/grupperingsvariabel, inte som en andra målvariabel.

```python
import pandas as pd
import numpy as np
from pathlib import Path

# Ändra sökvägen om CSV-filen ligger i en annan mapp.
DATA_PATH = Path("WVS_Cross-National_Wave_7_csv_v6_0.csv")

# Läs endast in variablerna som behövs för dagens första kontroll.
variables = [
    "A_YEAR", "B_COUNTRY", "B_COUNTRY_ALPHA",
    "Q49", "Q288", "Q262", "Q260", "Q275",
    "Q57", "Q58", "Q59", "Q1", "Q2", "Q3", "Q5"
]
df = pd.read_csv(DATA_PATH, usecols=variables, low_memory=False)

print(f"Ursprunglig datamängd: {df.shape[0]:,} rader och {df.shape[1]} valda kolumner")
print(f"Antal länder/regioner i hela filen: {df['B_COUNTRY_ALPHA'].nunique()}")
print("Undersökningsår:")
display(df["A_YEAR"].value_counts(dropna=False).sort_index().to_frame("Antal"))
```

```python
df
```

## 1. Preliminärt urval av europeiska länder/regioner

Följande 13 lands-/regionkoder används tills vidare. Urvalet är preliminärt och behöver diskuteras med läraren, särskilt hur Ryssland och Nordirland ska hanteras.

```python
europe_codes = [
    "AND", "CYP", "CZE", "DEU", "GBR", "GRC", "NIR",
    "NLD", "ROU", "RUS", "SRB", "SVK", "UKR"
]

df_europe = df[df["B_COUNTRY_ALPHA"].isin(europe_codes)].copy()

print(f"Preliminärt europeiskt urval: {len(df_europe):,} rader")
display(
    df_europe.groupby(["B_COUNTRY_ALPHA", "B_COUNTRY"])
    .size().reset_index(name="Antal respondenter")
    .sort_values("B_COUNTRY_ALPHA")
)
```

## 2. Kontrollera variabeltyper och värden

Viktigt: variablerna är numeriskt kodade, men alla ska inte behandlas som kontinuerliga tal. Flera är kategoriska eller ordinala. Koder för exempelvis ”vet inte” och ”inget svar” ska inte tolkas som riktiga mätvärden.

De specialkoder som kontrolleras nedan är preliminärt -1 till -5. Kontrollera den slutliga betydelsen i kodboken innan de omvandlas till saknade värden.

```python
special_codes = [-1, -2, -3, -4, -5]
analysis_vars = ["Q49", "Q288", "Q262", "Q260", "Q275",
                 "Q57", "Q58", "Q59", "Q1", "Q2", "Q3", "Q5"]

rows = []
for col in analysis_vars:
    s = df_europe[col]
    rows.append({
        "Variabel": col,
        "Datatyp": str(s.dtype),
        "NaN": int(s.isna().sum()),
        "Specialkoder (-1 till -5)": int(s.isin(special_codes).sum()),
        "Andel specialkoder (%)": round(s.isin(special_codes).mean() * 100, 2),
        "Antal unika värden": int(s.nunique(dropna=True)),
        "Min observerat värde": s.min(),
        "Max observerat värde": s.max()
    })

quality_overview = pd.DataFrame(rows)
display(quality_overview)
```

## 3. Kontrollera värdefördelningar

Granska frekvenserna för att upptäcka oväntade värden. Observera särskilt Q262 (ålder): värdet 99 behöver kontrolleras mot kodboken och får inte automatiskt antas vara en faktisk ålder eller en saknadkod. Q275 = 0 ska inte automatiskt tas bort; kontrollera koddefinitionen.

```python
for col in analysis_vars:
    print(f"\n--- {col} ---")
    display(
        df_europe[col].value_counts(dropna=False)
        .sort_index().rename_axis("Kod").to_frame("Antal")
    )
```

## 4. Saknade/specialkoder per land

Detta är särskilt viktigt för Q288 (hushållsinkomst). Eftersom inkomstskalan är relativ inom respektive land ska den inte tolkas som en gemensam absolut inkomstskala för hela Europa.

```python
missing_by_country = (
    df_europe.assign(**{
        col: df_europe[col].isin(special_codes)
        for col in analysis_vars
    })
    .groupby("B_COUNTRY_ALPHA")[analysis_vars]
    .mean()
    .mul(100)
    .round(1)
)
display(missing_by_country)
```

## 5. Skapa en preliminär analyskopia – ändra inte originaldata

När specialkoderna har bekräftats mot kodboken kan de ersättas med NaN i en separat analyskopia. Kör inte detta som ett slutgiltigt beslut innan koddefinitionerna har verifierats.

```python
# Preliminärt arbetssteg – verifiera specialkoderna först!
df_clean_candidate = df_europe.copy()
df_clean_candidate[analysis_vars] = df_clean_candidate[analysis_vars].replace(
    special_codes, np.nan
)

print("Rader före eventuell complete-case-filtrering:", len(df_clean_candidate))
print("\nAntal saknade värden efter preliminär kodning:")
display(df_clean_candidate[analysis_vars].isna().sum().to_frame("Saknade värden"))

# Vi tar INTE bort alla rader med minst ett saknat värde ännu.
# Beslut om imputation/complete-case görs efter att missingness har analyserats.
```
